### 3. Handling missing values (a simple approach) 

We have identified missing values, now, we want to handle them. One way is to use `dropna()` and remove all rows with missing values. That makes sens very often, but sometimes we would need to drop significant amount of laboroulsy gathered data, so we can try to handle these data in a better way - estimate them. 

**NOTE**: in practice this is questionable practice, but if we have a lot of missing data, there is nothing better we can do. 

`def handle_missing_values(df: pd.DataFrame) -> pd.DataFrame`

**Purpose**: Handle missing values appropriately

**Arguments**: 
- `df` (pd.DataFrame): Input DataFrame with missing values

**Returns**: 
- `pd.DataFrame`: DataFrame with missing values handled

**Requirements**: 
- For `Price`: Fill with the median price of the same `Category`
- For `Quantity`: Fill with the median quantity (if any missing)
- For categorical columns (`Region`, `Category`, `Product`): Fill with the mode or 'Unknown'
- Return the cleaned DataFrame


````python
### 3. Handling missing values 
def handle_missing_values(df: pd.DataFrame) -> pd.DataFrame:
    df[["Region", "Category", "Product"]] = df[["Region", "Category", "Product"]].fillna('Unknown')
    df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
    df["Price"] = df["Price"].fillna(df.groupby("Category")["Price"].transform("median"))
    return df

handle_missing_values(df).head()
````

### 4. Fixing data types 

def fix_data_types(df: pd.DataFrame) -> pd.DataFrame`

**Purpose**: Ensure all columns have correct data types

**Arguments**: 
- `df` (pd.DataFrame): Input DataFrame

**Returns**: 
- `pd.DataFrame`: DataFrame with corrected data types

**Requirements**: 
- Convert `Date` to datetime type
- Ensure `TransactionID` and `CustomerID` are integers
- Ensure `Quantity` is integer
- Ensure `Price` is float
- Ensure categorical columns are string type or category type
- Print the data types after conversion


````python
### 4. Fixing data types 

def fix_data_types(df: pd.DataFrame) -> pd.DataFrame:
    df["Date"]= pd.to_datetime(df["Date"])
    df[["TransactionID","CustomerID"]] =df[["TransactionID","CustomerID"]].astype("int32")
    df["Quantity"] = df["Quantity"].astype("int32")
    df["Price"] = df["Price"].astype("float32")
    df[["Region","Category","Product","MembershipTier","CustomerName"]] = df[["Region","Category","Product","MembershipTier","CustomerName"]].astype("string")
    return df.info()
fix_data_types(df)
````

### 5. Create total amount column 

`def add_total_amount(df: pd.DataFrame) -> pd.DataFrame`

**Purpose**: Create a new column for total transaction amount

**Arguments**: 
- `df` (pd.DataFrame): DataFrame with 'Quantity' and 'Price' columns

**Returns**: 
- `pd.DataFrame`: DataFrame with new 'Total_Amount' column added

````python
### 5. Create total amount column 

def add_total_amount(df: pd.DataFrame) -> pd.DataFrame:
    df["Total_Amount"] = df["Quantity"]*df["Price"]
    return df
add_total_amount(df)
````


### 7. Finding top customers

`def get_top_customers(df: pd.DataFrame, n: int = 5) -> pd.DataFrame`

**Purpose**: Find the top N customers by total purchase amount

**Arguments**: 
- `df` (pd.DataFrame): DataFrame with 'CustomerID', 'CustomerName', and 'Total_Amount'
- `n` (int): Number of top customers to return (default = 5)

**Returns**: 
- `pd.DataFrame`: DataFrame with columns ['CustomerID', 'CustomerName', 'Total_Purchase_Amount'] sorted in descending order

**Requirements**: 
- Group by CustomerID and CustomerName
- Sum the Total_Amount for each customer
- Sort by total purchase amount (descending)
- Return top N customers
- Round amounts to 2 decimal places

**Example Output**:
```
   CustomerID   CustomerName  Total_Purchase_Amount
0        1015    John Smith                25430.50
1        1023   Sarah Connor               22150.75
```

````python
### 7. Finding top customers

def get_top_customers(df: pd.DataFrame, n: int = 5) -> pd.DataFrame:
    df = add_total_amount(df) 
    df["Total_Purchase_Amount"] = round(df.groupby(["CustomerID","CustomerName"])["Total_Amount"].transform("sum"),2)
    df = df[["CustomerID","CustomerName","Total_Purchase_Amount"]].drop_duplicates().sort_values("Total_Purchase_Amount",ascending = False).head(n)
    return df

get_top_customers(df)
````

### 8. Average purchase per customer

```python
def calculate_avg_purchase_per_customer(df: pd.DataFrame) -> pd.DataFrame
```

**Purpose**: Calculate the average purchase amount per customer

**Arguments**: 
- `df` (pd.DataFrame): DataFrame with 'CustomerID', 'CustomerName', and 'Total_Amount'

**Returns**: 
- `pd.DataFrame`: DataFrame with columns ['CustomerID', 'CustomerName', 'Average_Purchase_Amount', 'Number_of_Transactions']

**Requirements**: 
- Group by CustomerID and CustomerName
- Calculate mean Total_Amount and count of transactions
- Sort by average purchase amount (descending)
- Round to 2 decimal places


````python
### 8. Average purchase per customer

def calculate_avg_purchase_per_customer(df: pd.DataFrame) -> pd.DataFrame:
    df = add_total_amount(df)
    df = df.groupby(["CustomerID", "CustomerName"]).agg(
        Average_Purchase_Amount=("Total_Amount", "mean"),
        Number_of_Transactions=("Total_Amount", "count")
    ).sort_values("Average_Purchase_Amount", ascending=False).reset_index()
    return df
calculate_avg_purchase_per_customer(df)
````

### 9. Finding best-selling product

```python
def get_best_selling_product(df: pd.DataFrame) -> tuple
```

**Purpose**: Find the product with the highest total quantity sold

**Arguments**: 
- `df` (pd.DataFrame): DataFrame with 'Product' and 'Quantity'

**Returns**: 
- `tuple`: (product_name, total_quantity_sold)

**Requirements**: 
- Group by Product
- Sum Quantity for each product
- Return the product with the highest total quantity

**Example Output**:
```python
('Smartphone', 245)
```


````python
### 9. Finding best-selling product

def get_best_selling_product(df: pd.DataFrame) -> tuple:
    product_sales = df.groupby('Product')['Quantity'].sum()
    best_product = product_sales.idxmax()
    max_quantity = product_sales.max()
    return (best_product, max_quantity)

get_best_selling_product(df)
````


### 10. Finding revenue per category 

```python
def calculate_revenue_by_category(df: pd.DataFrame) -> pd.DataFrame
```

**Purpose**: Calculate total revenue for each product category

**Arguments**: 
- `df` (pd.DataFrame): DataFrame with 'Category' and 'Total_Amount'

**Returns**: 
- `pd.DataFrame`: DataFrame with columns ['Category', 'Total_Revenue', 'Percentage_of_Total'] sorted by revenue (descending)

**Requirements**: 
- Group by Category
- Sum Total_Amount for each category
- Calculate percentage of total revenue
- Sort by Total_Revenue (descending)
- Round to 2 decimal places

**Example Output**:
```
       Category  Total_Revenue  Percentage_of_Total
0   Electronics       45230.50                 35.2
1      Clothing       32450.25                 25.3
```



````python
### 10. Finding revenue per category 

def calculate_revenue_by_category(df: pd.DataFrame) -> pd.DataFrame:
    df = add_total_amount(df)
    total_amount = df.groupby('Category').agg(
        Total_Revenue=("Total_Amount", "sum"))
    total = total_amount['Total_Revenue'].sum()
    total_amount['Percentage_of_Total'] = ((
        total_amount['Total_Revenue'] / total
    ) * 100).round(2)

    total_amount = total_amount.sort_values(by='Total_Revenue', ascending=False).reset_index()
    return total_amount

calculate_revenue_by_category(df)
````
