````python
import numpy as np
def average_grade(data_array):
    grades = data_array[1:, 2:].astype("float32")
    averages = np.mean(grades, axis=0).round(2)
    return averages.reshape(-1, 1)

average_grade(data)

def average_student_grade(data_array):
    grades = data_array[1:, 2:].astype(float)
    averages = np.mean(grades, axis=1)
    return averages.reshape(-1, 1)

average_student_grade(data)

def top_best_students(data_array):
    student_ids = data_array[1:, 0]
    grades = data_array[1:, 2:].astype(float)
    averages = np.mean(grades, axis=1).round(2)
    combined = np.column_stack((student_ids, averages))
    sorted_combined = combined[averages.argsort()[::-1]]
    top5 = sorted_combined[:5]
    return top5

top_best_students(data)
````

## Task 4: Calculating t-Tests

### Short Introduction to t-Test
A t-test is a statistical method used to determine if there is a significant difference between the means of two groups. It assumes the data follows a normal distribution and compares the means while accounting for variability (standard deviation) and sample size. The output includes a t-statistic (how different the means are) and a p-value (probability of observing the difference by chance). Typically, if p-value < 0.05, we assume that the difference is statistically significant (reject the null hypothesis of no difference).

You will do two tests:

1. Test if there is significant difference between two subjects: math and biology - perform t-test for those two subjects.
2. Verify if there is statistically significant difference between sexes:
  - Compute overall average grade per student (from Task 3).
  - Split into female (0) and male (1) groups using boolean masking.
  - Perform t-test between the average grades of females and males. Explain if there is statistically significant difference between sexes.

In order to do the t-test you will need to implement function `ttest(group1, group2)` that accepts two NumPy arrays with data for which we will do means comparision. This function should return a tuple with value of the t-test and p-value)

Implement the t-test using NumPy for the t-statistic formula (as shown below), and SciPy's `stats.t.cdf()` for p-value.

Example code snippet for t-test (adapt it!):

```python
import numpy as np
from scipy import stats

# For two samples: group1, group2 (1D arrays)
mean1 = np.mean(group1)
mean2 = np.mean(group2)
var1 = np.var(group1, ddof=1)
var2 = np.var(group2, ddof=1)
n1 = len(group1)
n2 = len(group2)
pooled_std = np.sqrt(((n1 - 1) * var1 + (n2 - 1) * var2) / (n1 + n2 - 2))
t_stat = (mean1 - mean2) / (pooled_std * np.sqrt(1/n1 + 1/n2))
df = n1 + n2 - 2

```

````python
import numpy as np
from scipy import stats

def ttest(group1, group2):
    mean1 = np.mean(group1)
    mean2 = np.mean(group2)
    
    var1 = np.var(group1, ddof=1)
    var2 = np.var(group2, ddof=1)
    
    n1 = len(group1)
    n2 = len(group2)
    
    pooled_std = np.sqrt(((n1 - 1) * var1 + (n2 - 1) * var2) / (n1 + n2 - 2))
    
    t_stat = (mean1 - mean2) / (pooled_std * np.sqrt(1/n1 + 1/n2))
    
    df = n1 + n2 - 2
    p_value = stats.t.cdf(-np.abs(t_stat), df) * 2
    
    return t_stat, p_value
````

````python
def test_gender(data_array):
    genders = data_array[1:, 1].astype(int)
    grades = data_array[1:, 2:].astype(float)
    
    averages = np.mean(grades, axis=1)
    
    female = averages[genders == 0]
    male = averages[genders == 1]
    
    return ttest(female, male)

test_gender(data)
````
