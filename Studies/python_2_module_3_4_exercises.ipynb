## Exercise 1: Series Practice

Create a Series containing the scores of 5 students with values [85, 92, 78, 95, 88].
1. Find students who scored above 85
2. Calculate the average score
3. Add 5 bonus points to all scores

````python
# Your code here
scores = pd.Series([85, 92, 78, 95, 88])

# 1. Students above 85
print(f"Students above 85: \n{scores[scores > 85]}")
# 2. Average score
print( f"Average score: {np.mean(scores).astype("float32")}")
# 3. Add bonus points
print( f"Average score:\n{scores+5}")

````

## Exercise 3: Missing Values Practice

Given the following dataset with missing values:
```python
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Math': [85, None, 78, 92, 88],
    'English': [90, 85, None, 88, 92],
    'Science': [None, 88, 82, 90, 85]
}
```

1. Create the DataFrame and identify missing values
2. Fill missing Math scores with the mean
3. Fill missing English scores with the median
4. Fill missing Science scores with 0
5. Calculate each student's average score

````python
# Your code here
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Math': [85, None, 78, 92, 88],
    'English': [90, 85, None, 88, 92],
    'Science': [None, 88, 82, 90, 85]
}

# 1. Create DataFrame and identify missing values
df = pd.DataFrame(data)
df.isnull()
# 2-4. Fill missing values
df["Math"] = df["Math"].fillna(df["Math"].mean())
df['English'] = df['English'].fillna(df['English'].median())
df['Science'] = df['Science'].fillna(0)

# 5. Calculate average scores
df["Average"] = df[['Math','English','Science']].mean(axis = 1)
df[["Student","Average"]]

# Hint: use mean(axis=1) for subjects columns, axis = 1 means horizontal directrion,
# so this is what we want
````
