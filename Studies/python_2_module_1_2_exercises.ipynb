## Exercise 1 (Python review)

Our task is to do some operations on the made-up products database

````python
# We store product database ID, SKU, price, category, stock amount
products = [(1, 'Wireless Mouse', 'WMA-123', 29.99, 'Electronics', 150),
            (2, 'Keyboard', 'KBD-200', 129.99, 'Electronics', 200),
            (3, 'Yamaha Classical Guitar', 'CGT-467', 550.99, 'Music', 190),
            (4, 'Gibson Les Paul', 'FLP-467', 2000.0, 'Music', 3),
            (5, 'Fender Stratocaster', 'FST-111', 3000.0, 'Music', 50),
            (6, 'Sony Trinitron Color CRT TV | Vintage', 'KV-1380R 13', 368.0, 'Electronics', 5)]
````

1. Create function `avarage_price()` that will return avarage products price
2. Create function `categories()` that will return all product categories
3. Create function `min_max()` that will return minimal and maximal product prices.

````python
df = pd.DataFrame(products)
df.columns=["ID","Product", "SKU", "price", "category", "stock amount"]
````

````python
def avarage_price(x=df["price"]):
    return np.mean(x).round(2)
print(f"1. Average price: {avarage_price()}")
def categories(x=df["category"]):
    return list(set(x))
print(f"2. Categories of products: {categories()}")
def min_max(x=df["price"]):
    return f"Min: {np.min(x)} \n Max: {np.max(x)}"
print(f"3. {min_max()}")
````
