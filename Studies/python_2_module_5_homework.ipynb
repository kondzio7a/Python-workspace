## What's that Bayes thingy?

**(Naive) Bayes** is a family of classification methods based on a very simple idea:

> *If I know how often something happens in general, and how often it happens when certain conditions are present, I can update my belief when I observe those conditions.*

In other words, Naive Bayes answers questions like:

* Given these symptoms, how likely is this disease?
* Given these words, how likely is this message spam?
* What is probability of cancer, given test results?

BTW: don't worry about the "naive" meaning, we will get to that later.

It is a **probabilistic classifier**: instead of directly assigning a label (healthy/not healthy), it computes probabilities and chooses the most likely label (called in this case "class").

So, Bayes Rule calculates something called **posterior probability** - our belief in something *after* seeing the data.

$$
P(\text{class} \mid \text{observations})
$$

We will look on the typical example (before we move to our spam detection case) of cancer detection.

So, reasoning goes like this: we know that say, 1%, of population has a certain cancer. That's great that we have such statistics from our NFZ, but what does it mean for me? Depending if I am an optimist or pessimist I can interpret this differently ;). 

So, maybe, we can do better. For instance we can do some test. But no test is perfect, so still we need to estimate somehow what is the probability of being sick given the test is positive.

And this is exactly what we use Bayes rule for...

## Naive Bayes Rule - Medical Test Example

### Bayes' Rule

The mathematical form of Bayes’ rule is:

$$
P(H|E) = \frac{P(E|H) \, P(H)}{P(E)}
$$

where:

- $H$ - hypothesis (person has cancer)  
- $E$ - evidence (test is positive)
- $P(H)$ - prior probability, it tells us what is probability that a randomly chosen person has cancer - this is fairly easy to estimate, we have statistics what percentage of population is affected by a given cancer.
- $P(E)$ - evidence probability, that is how many tests gave good or bad results - again, this is easy to count, we just need to do a test for a lot of people and check whether or not it gave proper result. 
- $P(H|E)$ is a conditional probability that person has cancer given the test results. Note: in general this is hard to measure, a person comes to the test center, we do test, we get the result and... What's then? But there is something that we can estimate...
- ... which is $P(E|H)$ - this is conditional probability that test gave positive result given a person is sick (what we know otherwise, because cancer was detected in some other way). We just need to test a lot of people using our test and we can estimate how often test was correct (or not) if we had a person who was really sick.

So Bayes rule allows exchanging calculation of something difficult to measure for something that we can somehow calculate.

Let's look on the example.

#### Assumptions about the test and cancer

- 1% of population have a certain cancer
- The test is:
  - 99% positive if the person has cancer
  - 5% positive if the person is healthy (false positive)

#### Question

If the test is positive, what is the probability that the person actually has cancer?


#### Step 1: Prior Probability

The **prior probability** represents the base rate of cancer in the population.

$$
P(H) = P(\text{cancer}) = 0.01
$$

This means that before observing the test result, the probability that a randomly selected person has cancer is **1%**.


#### Step 2: Likelihood

The **likelihood** represents the probability that the test is positive if the person actually has cancer.

$$
P(E|H) = P(\text{positive}|\text{cancer}) = 0.99
$$

This is the **true positive rate (test sensitivity)** and is obviously easy to verify.


#### Step 3: Evidence Probability

The probability of observing a positive test overall is calculated from the example population.

Out of **10,000 people**:

- 100 have cancer  
  - 99 test positive
- 9,900 are healthy  
  - 495 test positive (false positives)

Total positive tests:

$$
99 + 495 = 594
$$

Therefore:

$$
P(E) = P(\text{positive}) = \frac{594}{10000} = 0.0594
$$

This is also easy to calculate, we just test a lot of people and write down if the test gave proper result or not.

#### Step 4: Posterior Probability

Now, this is the part that is hard to estimate and that's why Bayes' Rule is so useful.

The **posterior probability** is the probability of cancer given that the test result is positive.

$$
P(H|E) = P(\text{cancer}|\text{positive})
$$

From the counts:

$$
P(H|E) = \frac{99}{594} \approx 0.167
$$

So the probability that a person actually has cancer given a positive test is:

$$
P(\text{cancer}|\text{positive}) \approx 16.7\%
$$

#### Summary of Bayes Rule Components

| Bayes term | Meaning | Value |
|------------|---------|------|
| $P(H)$ | Prior probability of cancer | 1% |
| $P(E|H)$ | Probability of positive test if cancer | 99% |
| $P(E)$ | Overall probability of positive test | 5.94% |
| $P(H|E)$ | Probability of cancer given positive test | ≈16.7% |


#### Key Insight

Even a very accurate test can produce many **false positives** when the disease is rare.  
Because cancer occurs in only 1% of the population, the number of healthy people is very large, which leads to many false positive results.

This is a classic illustration of **base rate effects** and why **Bayesian reasoning** is important in interpreting medical tests.

You make say that those probabilities are made up and are artificially high, so in reality there is no that much false positives.

It turns out that is the opposite. For instance mammography, used to detect breast cancer has false positive rate 10%-12% and true positive rate (sensitivity) is 87%. 

## What "naive" means?

Naive Bayes is a simplification of "full blown Bayes". It makes an assumption that various conditions that contribute to the final probability are independent. So, for instance, we assume that various symptomps of having a cancer are independent (which is not 100% true, obviously). 

Similarly, when we do spam detection, we assume that appearance of certain words contributing to a message being classified as spam are also independent. Again, sometimes it is true, but sometimes it is not, for instance "win" and "prize" will often show up together in spam messages, so their appearance is not independent on each other. 

Nevertheless, "naive" assumption typically works.

## Spam detection

Now let’s apply the same logic to text.

### Goal:

Classify a message as **spam** or **ham**

### Features:

Words in the message:

* “free”
* “win”
* “meeting”
* “tomorrow”

---

## Step 1: Learn from data

From many labeled messages, we estimate:

* How common spam is overall
  - ( P(\text{spam}) )

* How often each word appears in spam vs ham
  - ( P(\text{“free”} \mid \text{spam}) )
  - ( P(\text{“free”} \mid \text{ham}) )

These are just **counts normalized into probabilities**.

## Step 2: Classify a new message

Message:

> “free prize tomorrow”

Naive Bayes asks two competing questions:

> How likely is this message if it were spam?

> How likely is this message if it were ham?

It computes:

* Spam likelihood:

  * spam is common?
  * “free” common in spam?
  * “prize” common in spam?
  * “tomorrow” common in spam?

* Ham likelihood:

  * ham is common?
  * same words, but with ham statistics

The class with the **higher probability wins**.

First we quickly go through initial steps that are not dependent on model used and that lead us to the derivation of Bag of Words matrix.

```python
import numpy as np
import pandas as pd

# Our dataset

data = [
    ("ham",  "Hey, are we still meeting tomorrow?"),
    ("ham",  "Can you send me the lecture notes?"),
    ("ham",  "I will call you in 10 minutes."),
    ("ham",  "Thanks for your help today."),
    ("ham",  "Are you coming to class?"),
    ("ham",  "Let's study together this weekend."),
    ("ham",  "Please remember the meeting at 9am."),
    ("ham",  "See you soon."),
    ("ham",  "Happy birthday! Have a great day."),
    ("ham",  "Don't forget to bring the book."),

    ("spam", "WIN cash now! Click here to claim your prize."),
    ("spam", "Congratulations! You won a free gift card."),
    ("spam", "Free entry in weekly competition, text WIN now."),
    ("spam", "Earn money fast from home, limited offer."),
    ("spam", "URGENT! Your account needs verification now."),
    ("spam", "Winner! Call now to collect reward."),
    ("spam", "Exclusive discount available today only."),
    ("spam", "Cheap meds online, buy now."),
    ("spam", "Claim your free prize immediately."),
    ("spam", "Last chance to win big cash reward."),
]

df = pd.DataFrame(data, columns=["label", "text"])
df["y"] = (df["label"] == "spam").astype(int)

# Tokenization

def tokenize(text):
    """
    Convert text to lowercase, remove punctuation,
    and split into tokens.
    """
    text = text.lower()
    cleaned = []
    for ch in text:
        if ch.isalnum() or ch.isspace():
            cleaned.append(ch)
        else:
            cleaned.append(" ")
    return [tok for tok in "".join(cleaned).split() if tok]


# Example
print(tokenize("WIN cash now! Click here."))

# ----------------------------
# 3) Build vocabulary
# ----------------------------
def build_vocab(texts, min_count=1):
    """
    Build a vocabulary dictionary: token -> index
    """
    tokens = pd.Series(texts).apply(tokenize)
    all_tokens = tokens.explode().dropna()
    counts = all_tokens.value_counts()

    vocab_tokens = counts[counts >= min_count].index.tolist()
    vocab = {tok: i for i, tok in enumerate(vocab_tokens)}
    return vocab, counts

vocab, token_counts = build_vocab(df["text"], min_count=1)

print("Vocabulary size:", len(vocab))
token_counts.head(15)

def vectorize_bow(texts, vocab):
    """
    Convert list of texts into bag-of-words matrix
    X shape = (n_samples, vocab_size)
    """
    n = len(texts)
    V = len(vocab)
    X = np.zeros((n, V), dtype=np.float64)

    for i, text in enumerate(texts):
        for tok in tokenize(text):
            j = vocab.get(tok)
            if j is not None:
                X[i, j] += 1.0
    return X

X = vectorize_bow(df["text"].tolist(), vocab)
y = df["y"].to_numpy()

print("Features (messages) shape", X.shape) 
print("Target variable spam/ham shape", y.shape)
```

```python
# Train/test split
def train_test_split(X, y, test_size=0.3, seed=42):
    rng = np.random.default_rng(seed)
    idx = np.arange(len(y))
    rng.shuffle(idx)

    n_test = int(np.ceil(test_size * len(y)))
    test_idx = idx[:n_test]
    train_idx = idx[n_test:]

    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, seed=7)

X_train.shape, X_test.shape
```

```python
# ----------------------------
# 6) Train Multinomial Naive Bayes
# ----------------------------
def fit_multinomial_nb(X, y, alpha=1.0):
    """
    Train a Multinomial Naive Bayes model.

    Returns a dictionary containing:
    - classes
    - log prior probabilities
    - log conditional probabilities of features given class
    """
    classes = np.unique(y)
    n_features = X.shape[1]

    class_log_prior = {}
    feature_log_prob = {}

    for c in classes:
        X_c = X[y == c]

        # Prior probability of class c
        prior = X_c.shape[0] / X.shape[0]
        class_log_prior[c] = np.log(prior)

        # Count total occurrences of each word in class c
        word_counts = X_c.sum(axis=0)

        # Laplace smoothing
        smoothed_counts = word_counts + alpha
        smoothed_total = smoothed_counts.sum()

        # Conditional probabilities P(word | class)
        feature_log_prob[c] = np.log(smoothed_counts / smoothed_total)

    model = {
        "classes": classes,
        "class_log_prior": class_log_prior,
        "feature_log_prob": feature_log_prob,
        "alpha": alpha,
        "n_features": n_features
    }

    return model
```

```python
model = fit_multinomial_nb(X_train, y_train, alpha=1.0)
model
```

```python
# Predict class probabilities in log-space
def predict_log_proba_nb(X, model):
    """
    Compute log-probabilities (up to a constant) for each class.
    Returns array of shape (n_samples, n_classes)
    """
    classes = model["classes"]
    scores = np.zeros((X.shape[0], len(classes)), dtype=np.float64)

    for j, c in enumerate(classes):
        log_prior = model["class_log_prior"][c]
        log_feature_prob = model["feature_log_prob"][c]

        # For multinomial NB:
        # log P(class) + sum_j x_j * log P(word_j | class)
        scores[:, j] = log_prior + X @ log_feature_prob

    return scores
```

```python
# ----------------------------
# 8) Predict labels
# ----------------------------
def predict_nb(X, model):
    """
    Predict class labels for X
    """
    scores = predict_log_proba_nb(X, model)
    class_indices = np.argmax(scores, axis=1)
    return model["classes"][class_indices]

y_pred = predict_nb(X_test, model)
y_pred
```

```python
# ----------------------------
#9) Predict new messages
# ----------------------------
def predict_text_nb(text, vocab, model):
    X_new = vectorize_bow([text], vocab)
    pred = predict_nb(X_new, model)[0]
    return "spam" if pred == 1 else "ham"

examples = [
    "Reminder: our class starts at 10am tomorrow.",
    "Congratulations! Claim your free cash prize now.",
    "Can you send me the homework?",
    "WIN a free phone today, click now!"
]

for text in examples:
    label = predict_text_nb(text, vocab, model)
    print(f"Text: {text}")
    print(f"Prediction: {label}")
    print("-" * 60)    
```

## Appendix: Bayes' Theorem: Mathematical Explanation and Intuition

This is for those who like to dig deeper and are not scared of math. BTW: the math here is not scary, so give it a chance!

As we know, Bayes' theorem is a funky way of translating a difficult task to some more managable tasks, so we can do some trick that exchanges calculating one conditional probability for calculating another conditional probability, which is supposed to be easier.

But how we know this works? In other words, we want to proof Bayes theorem. Sounds difficult, but it isn't, we will just use a secret tool that mathematicians are using - if we need to proof something, we just need to calculate it. This secret, for some reason, is not revealed when we are at school, that's why many pupils and students are scared of proofs.

### Conditional Probability

To understand Bayes' theorem, we start with the definition of **conditional probability**.

The probability of event $A$ given that event $B$ occurred is defined as:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

where:

- $P(A \cap B)$ is the probability that both events occur
- $P(B)$ is the probability that event $B$ occurs

But why is that? Well, we know that B already has happened, so we know that this is the whole space of events we have at hand. So the probability that A and B happens is divided by that what we have - that is B.

Similarly, we can define conditional probability for the opposite case - probability of $B$ happening, given $A$ happened:

$$
P(B \mid A) = \frac{P(A \cap B)}{P(A)}
$$

Both formulas use the same joint probability $P(A \cap B)$, so we can do another secret mathematicians trick - we calculate the same thing twice and compare it...


### Deriving Bayes' Theorem

From the definition above:

$$
P(A \cap B) = P(A \mid B)P(B)
$$

But also:

$$
P(A \cap B) = P(B \mid A)P(A)
$$

Since both expressions equal the same quantity, we can equate them:

$$
P(A \mid B)P(B) = P(B \mid A)P(A)
$$

Solving for $P(A \mid B$) gives **Bayes' theorem**:

$$
P(A \mid B) =
\frac{P(B \mid A)P(A)}{P(B)}
$$

### Interpretation of the Terms

For completness, let's recall that what we have already covered:
each component of Bayes' theorem meaning.

$$
P(A \mid B) =
\frac{P(B \mid A)P(A)}{P(B)}
$$

- $P(A)$ — prior probability  
  Initial belief about event $A$ before observing data.

- $P(B \mid A)$ — likelihood  
  Probability of observing evidence $B$ if hypothesis $A$ is true.

- $P(B)$ — evidence (normalization factor)  
  Overall probability of observing $B$.

- $P(A \mid B)$ — posterior probability  
  Updated belief about $A$ after observing $B$.


### Connection to Naive Bayes

Now let's see how this naivness assumption manifests in mathematical terms.

Naive Bayes classifiers apply Bayes' theorem to classification problems: for a class $C$ and features $x_1, x_2, \dots, x_n$, the classifier computes:

$$
P(C \mid x_1, x_2, ..., x_n)
$$

Class means a "label" that is value we want to predict (like spam/ham), features are predictors that we use to predict an outcome - in case of spam example those are messages.

Using Bayes' rule:

$$
P(C \mid x_1, ..., x_n) =
\frac{P(x_1, ..., x_n \mid C) P(C)}
{P(x_1, ..., x_n)}
$$

The **Naive Bayes assumption** simplifies the likelihood by assuming conditional independence:

$$
P(x_1, ..., x_n \mid C) =
\prod_{i=1}^{n} P(x_i \mid C)
$$

This simplification makes the computation efficient even for very high-dimensional data such as text.
